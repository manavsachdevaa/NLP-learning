```python
#TOkennization
from nltk.tokenize import word_tokenize
#Stopwords List
from nltk.corpus import stopwords
#Stemming
from nltk.stem import PorterStemmer
#Lemmatization
from nltk.stem import WordNetLemmatizer
import string
import joblib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

import nltk
nltk.download("punkt")     #punkt-> punctuation
nltk.download('punkt_tab')
nltk.download("stopwords")
nltk.download("wordnet")
```

```python
def preprocess_text(documents):
  final_list=[]
  for sent in documents:
    sent=sent.lower()
    tokens=word_tokenize(sent)
    english_stopwords=stopwords.words("english")
    filtered_tokens=[]

    for token in tokens:
      if token not in english_stopwords:
        filtered_tokens.append(token)
    punctuations=string.punctuation
    clean_tokens=[]
    for token in filtered_tokens:
      if token not in punctuations:
        clean_tokens.append(token)
    wnet=WordNetLemmatizer()
    lemmtized_words=[]
    for word in clean_tokens:
      lemmtized_words.append(wnet.lemmatize(word,"v"))
    alpha_list=[]
    for word in lemmtized_words:
      if word.isalpha():
        alpha_list.append(word)
    final_list.append(" ".join(alpha_list))
  return final_list
```

```python
model=joblib.load("intent_model.pkl")
vectorizer=joblib.load("tfidf_vec.pkl")
```

```python
user_msg="open youtube"  #Testing data
processed=preprocess_text([user_msg])
user_vector=vectorizer.transform(processed)    #For testing only transform for training fit_transform
prediction=model.predict(user_vector)
print("Prediction is: ",prediction)
```
