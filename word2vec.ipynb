```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')
```

```python
sentences=[
    "I am learning natural language processing",
    "Natural language processing is a part of artificial intelligence",
    "Artificial intelligence is a part of computer science",
    "I am also learning machine learning",
    "I also want to learn Deep learning",
    "Deep Learning is a very popular programming language",
    "Neural networks are popular"
]
```

```python
from nltk.tokenize import word_tokenize
# tokenized_sentences = [word_tokenize(s.lower()) for s in sentences]
tokenized_sentences=[]
for s in sentences:
  tokenized_sentences.append(word_tokenize(s.lower()))
print(tokenized_sentences)
```

```python
model=Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,   #embedding dimension- dimension of word vectors
    window=3,          #context window size- max distance b/w target and context windw
    min_count=1        #keep all words - ignore words appearing less than this

)
```

```python
print(model.wv.similarity('artificial','intelligence'))
```

```python
print(model.wv.similarity('deep','learning'))
```

```python
model.wv.most_similar('deep')
```

```python
print(model.wv.most_similar('deep',topn=5))
```
